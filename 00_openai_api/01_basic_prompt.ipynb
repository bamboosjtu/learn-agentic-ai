```python
import os
from openai import OpenAI
import rich
```

```python
API_KEY = os.environ.get('OPENAI_API_KEY')
BASE_URL = os.environ.get('OPENAI_BASE_URL')
MODEL = "gpt-5.4"

openai = OpenAI(api_key=API_KEY, base_url=BASE_URL)
```

# Chat Completion API

https://platform.openai.com/docs/quickstart?api-mode=chat

```python
prompts = [
    { "role": "user", "content": "讲一个关于互联网的冷笑话"}
]

response = openai.chat.completions.create(
    model=MODEL,
    messages= prompts
)
```

```python
rich.print(response)
```

```python
rich.print(response.choices[0])
print()
rich.print(response.choices[0].finish_reason)
```

```python
rich.print(response.choices[0].message)
```

```python
print(response.choices[0].message.function_call)
print(response.choices[0].message.tool_calls)
print(response.choices[0].message.annotations)
```

```python
print(response.choices[0].message.content)
```

```python
print(response.choices[0].message.reasoning_content)
```

# Responses API

https://platform.openai.com/docs/quickstart?api-mode=responses

```python
response = openai.responses.create(
    model=MODEL,
    input="讲一个关于互联网的冷笑话。" # Note: Now we can send user prompt directly as string to input property
)
```

```python
print(response.output_text) # Simple way to access the response text from model
```

```python
rich.print(response)
```

```python
print(response.tool_choice)
print(response.tools)
print(response.status)
print(response.store)
```

```python
rich.print(response.output[-1])
```
